## Cargar librerias y el archivo

### Descripción Breve de las Columnas 
- job_title_short: Nombre corto del puesto de trabajo.
- job_title: Nombre más extenso del puesto de trabjo, es el nombre que le coloca la empresa al publicar la oferta.
- job_location: Ubicación física donde se desarrollara el trabajo. 
- job_via: Plataforma donde se publico la oferta de trabajo.
- job_schedule_type: Forma en la que se va a trabajar 'tiempo completo', 'media jornada', etc. 
- search_location: Lugar en donde se buscan trabajadores. 
- job_posted_date: Fecha de publicación de la oferta de trabajo.
- job_no_degree_mention: Si se solicita o no se solicita titulo en la oferta de trabajo.
- job_health_insurance: Si se ofrece seguro medico. 
- job_country: País donde se buscan trabajadores. 
- salary_rate: Tasa salarial, 'hora','año','mes', etc. 
- salary_year_avg: Salario promedio por año.
- salary_hour_avg: Salario promedio por hora.
- company_name: Nombre de la compañia que realiza la oferta. 
- job_skills: Lista con las habilidades que se solicitan para el puesto 
- job_type_skills: Diccionario con la clasificación de tipos de habilidades por ejemplo: 'analyst_tools', 'programming', etc.

```python
import pandas as pd 
from rutas import ruta_archivo_limpio
import matplotlib.pyplot as plt 
import seaborn as sns 

df = pd.read_parquet(ruta_archivo_limpio())
df_trabajar = df.copy()
df_trabajar
```

## Análisis Exploratorio de Datos

### Top 10 Puestos de Trabajo con más Ofertas

```python
# Seleccion de los 10 puestos con mayores ofertas 
top_10_puestos = df_trabajar.job_title_short.value_counts().head(10).to_frame()

# Grafico y modificacion del aspecto visual 
sns.set_theme(style='whitegrid')
sns.barplot(
    data=top_10_puestos,
    x='count',
    y='job_title_short',
    hue='count',
    palette='dark:b_r',
    legend=False)
ax = plt.gca()
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x,pos: f"{round(x/1000)}k"))
sns.despine()
plt.title('¿Que Puestos de Trabajo Tienen Mayor Demanda?',fontsize = 15,pad=10)
plt.ylabel('')
plt.xlabel('Cantida de Ofertas')
plt.show()
```

### Top 10 Países con Mayor Cantidad de Ofertas de Trabajo 

```python
# Seleccion de los 10 puestos con mayores ofertas 
top_10_paises = df_trabajar.job_country.value_counts().head(10).to_frame()

# Grafico y modificacion del aspecto visual 
sns.set_theme(style='whitegrid')
sns.barplot(
    data=top_10_paises,
    x='count',
    y='job_country',
    hue='count',
    palette='dark:b_r',
    legend=False
)
ax = plt.gca()
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x,pos: f"{round(x/1000)}k"))
sns.despine()
plt.title('¿Que Paises Publican Mayor Cantidad de Ofertas de Trabajo?',fontsize = 15,pad=10)
plt.ylabel('')
plt.xlabel('Cantida de Ofertas')
plt.show()
```

### Top 10 Empresas con Mayor Cantidad de Ofertas de Trabajo 

```python
# Seleccion de los 10 puestos con mayores ofertas 
top_10_empresas = df_trabajar.company_name.value_counts().head(10).to_frame()

# Grafico y modificacion del aspecto visual 
sns.set_theme(style='whitegrid')
sns.barplot(data=top_10_empresas,
            x='count',
            y='company_name',
            hue='count',
            palette='dark:b_r',
            legend=False
)
sns.despine()
plt.title('¿Que Empresas Publican Mayor Cantidad de Ofertas de Trabajo?',fontsize = 15,pad=10)
plt.ylabel('')
plt.xlabel('Cantida de Ofertas')
plt.show()
```

### Análisis de Beneficios y Requisistos Laborales (Teletrabajo, solicita titulo, ofrece seguro de salud)

```python
dic_columnas_titulo = {
    'job_work_from_home': '¿Permite Teletrabajo?',
    'job_no_degree_mention': '¿No Menciona Titulo?',
    'job_health_insurance': '¿Ofrece Seguro Medico?'
}
fig, ax = plt.subplots(1,3,figsize= (15,10))

for i, (columna,titulo) in enumerate(dic_columnas_titulo.items()):
    analisis_beneficios = df_trabajar[columna].value_counts()
    
    # Etiquetas de Verdadero/Falso (Porcentaje)
    etiquetas = [
        f"{etiqueta}\n({round(analisis_beneficios.loc[etiqueta]/analisis_beneficios.sum()*100,2)}%)"
        for etiqueta in analisis_beneficios.index
    ]
    analisis_beneficios.plot(kind='pie',ax=ax[i],labels=etiquetas )
    ax[i].set_title(titulo,fontsize=20)
    ax[i].set_ylabel('')
```
